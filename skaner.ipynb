```python
import re

with open('wyrazenie.txt', 'r', encoding='utf-8') as file:
    ###
    nawias=0
    ###
    token=""
    wartosc=""
    moja_lista = []
    i=0
    while True:
        #
        i+=1
        #print(i)
        znak = file.read(1)
        #błędy
        if znak.isspace():
            continue
        if wartosc in "+\-*/" and znak in "+\-*/":
            raise Exception(f"Znaki działania nie moga byc obok siebie! kolumna: {i}")
        

        ###
        if token in ("liczba", "identyfikator", "nawias zamkniecie")  and znak =="(":
            if token=="liczba":
                raise Exception(f"brak znaku miedzy liczba a nawiasem otweirajacym! kolumna: {i}")
            if token=="identyfikator":
                raise Exception(f"brak znaku miedzy identyfikatorem a nawiasem otweirajacym! kolumna: {i}")
            if token=="nawias zamkniecie":
                raise Exception(f"brak znaku miedzy nawiasem otwierajacym a nawiasem otweirajacym! kolumna: {i}")
        if token =="nawias zamkniecie" and znak not in " +\-*/)":
            if znak in"123456789":
                raise Exception(f"brak znaku miedzy nawiasem zamykajacym a liczba! kolumna: {i}")
            else:
                raise Exception(f"brak znaku miedzy nawiasem zamykajacym a identyfikatorem! kolumna: {i}")
        if wartosc in "+\-*/" and znak ==")":
            raise Exception(f"brak liczby lub identyfikatora miedzy znkaiem a liczba! kolumna: {i}")
        ###
            
        
        if re.fullmatch(r"[0-9]", znak) and token !="identyfikator":
            if token != "liczba" and token!="":
                moja_lista.append({token: wartosc})
                wartosc=""
            token="liczba"
            wartosc += znak
        elif znak not in r"+\-*/()":
            ###
            if token != "identyfikator" and token != "liczba" and token!="":
                moja_lista.append({token: wartosc})
                wartosc=""
            ###
            token="identyfikator"
            wartosc += znak
        elif znak == "+"and token!="":
            if token != "plus":
                moja_lista.append({token: wartosc})
            token="plus"
            wartosc="+" 
        elif znak == "-":
            if token != "odejmowanie":
                moja_lista.append({token: wartosc})
            token="odejmowanie"
            wartosc="-"
        elif znak == "*"and token!="":
            if token != "mnozenie":
                moja_lista.append({token: wartosc})
            token="mnozenie"
            wartosc="*"
        elif znak == "/"and token!="":
            if token != "dzielenie":
                moja_lista.append({token: wartosc})
            token="dzielenie"  
            wartosc="/"
        elif znak == "(":
            if token != "nawias otwarcie":
                moja_lista.append({token: wartosc})
            token="nawias otwarcie" 
            wartosc="("
            ###
            nawias+=1
            ###
        elif znak == ")"and token!="":
            ###
            if nawias==0:
                raise Exception("Brak otwarcia nawiasu")
            ###
            if token != "nawias zamkniecie":
                moja_lista.append({token: wartosc})
            nawias-=1
            token="nawias zamkniecie"
            wartosc=")"
        elif  znak=="":
            moja_lista.append({token: wartosc})
            
            ###

            if(nawias!=0):
                raise Exception("brak zamkniecia nawiasu")
            ###
            break
print(moja_lista)
#2+3aaa*(76+8/3)+ 3*(9-3)
#2+3*(76+8/3)+ 3*(9-3)
#2+3a2a2a*(76+8/3)++ 3*(9-3)

```
