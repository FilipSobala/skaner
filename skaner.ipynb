```python
import re

with open('wyrazenie.txt', 'r', encoding='utf-8') as file:
    
    token=""
    wartosc=""
    moja_lista = []
    while True:
        znak = file.read(1)
        #błędy
        if wartosc in "+\-*/()" and znak in "+\-*/()":
            raise Exception("Znaki działanie nie moga byc obok siebie!")
        
        
        
        if re.fullmatch(r"[0-9]", znak) and token !="identyfikator":
            if token != "liczba" and token!="":
                moja_lista.append({token: wartosc})
                wartosc=""
            token="liczba"
            wartosc += znak
        elif znak not in r"+\-*/()":
            if token != "identyfikator" and token!="":
                moja_lista.append({token: wartosc})
                wartosc=""
            token="identyfikator"
            wartosc += znak
        elif znak == "+"and token!="":
            if token != "plus":
                moja_lista.append({token: wartosc})
            token="plus"
            wartosc="+" 
        elif znak == "-":
            if token != "odejmowanie":
                moja_lista.append({token: wartosc})
            token="odejmowanie"
            wartosc="-"
        elif znak == "*"and token!="":
            if token != "mnozenie":
                moja_lista.append({token: wartosc})
            token="mnozenie"
            wartosc="*"
        elif znak == "/"and token!="":
            if token != "dzielenie":
                moja_lista.append({token: wartosc})
            token="dzielenie"  
            wartosc="/"
        elif znak == "(":
            if token != "nawias otwarcie":
                moja_lista.append({token: wartosc})
            token="nawias otwarcie" 
            wartosc="("           
        elif znak == ")"and token!="":
            if token != "nawias zamkniecie":
                moja_lista.append({token: wartosc})
            token="nawias zamkniecie"
            wartosc=")"
        elif  znak=="":
            moja_lista.append({token: wartosc})
            break
print(moja_lista)
#2+3aaa*(76+8/3)+ 3*(9-3)

```
